Operators in Python

```python
# Arithmetic Operators
# Examples:
a = 10
b = 5
print(a + b)  # Output: 15  
print(a - b)  # Output: 5
print(a * b)  # Output: 50
print(a / b)  # Output: 2.0
print(a // b)  # Output: 2 (floor division)
print(a % b)  # Output: 0 (modulo)
```

```python
#Comparison Operators
# Examples:
x = 10
y = 5
print(x > y)  # Output: True #greater than
print(x < y)  # Output: False #less than
print(x == y)  # Output: False #equal to
print(x != y)  # Output: True #not equal to
print(x >= y)  # Output: True #greater than or equal to
print(x <= y)  # Output: False #less than or equal to
```

```python
#Logical Operators
# Examples:
p = True
q = False
print(p and q)  # Output: False (logical AND)
print(p or q)   # Output: True (logical OR)
print(not p)    # Output: False (logical NOT)
```

```python
#Precedence of Operators
# Examples:
x = 10
y = 5
z = 2
print(x + y * z)  # Output: 20 (multiplication is performed before addition)
print((x + y) * z)  # Output: 30 (parentheses override the default precedence)

#left to right associativity
# Examples:
print(x - y - z)  # Output: 3 (subtraction is performed left to right)
eq = 10 + 5 / 2 - 3 * 4
#here, the division is performed first, followed by multiplication, addition, and subtraction, resulting in 10 + 2.5 - 12 = 0.5
```
